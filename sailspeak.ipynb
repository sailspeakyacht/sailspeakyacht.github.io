# SailSpeak

Official site: [sailspeak.com](https://sailspeak.com/)

Captains’ network for yacht and boat rentals. You book on the captain’s page. Payment is held until the charter ends. If the owner cancels, another boat or a refund. Reviews go both ways after the trip.

This notebook is a public brand note, not a booking form.

```python
SAILSPEAK = {
    "site": "https://sailspeak.com/",
    "captains": "https://sailspeak.com/make-money/",
    "thailand": "https://sailspeak.com/thailand/",
    "phuket": "https://sailspeak.com/phuket-thailand/",
    "pattaya": "https://sailspeak.com/pattaya-thailand/",
}
ports = [
    ("Phuket", "Coral / Maiton / Khai / Phi Phi / Racha, fishing"),
    ("Pattaya", "island hops, fishing"),
    ("Koh Samui", "catamaran, squid fishing, Koh Taen"),
    ("France / Monaco", "day motor yachts"),
]
print("SailSpeak — book with the captain")
print(SAILSPEAK["site"])
print()
for port, note in ports:
    print(f"  {port:16}  {note}")
```

## Hold, rating, captains

- Money stays on the site until the rental period is over.
- Guest and captain review each other. Repeated cancellations push listings down or off search.
- A captain can add a boat in a few minutes and get a public page: [make-money](https://sailspeak.com/make-money/).
